# Face data 

The file `faces.mat` contains a single array `face` of size $24\times 21\times 600$: 600 grayscale face images of size $24\times 21$ pixels, three images per subject for 200 subjects (images $3i, 3i+1, 3i+2$ belong to subject $i$).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

data = loadmat('faces.mat')
face = data['face']
print('face array shape:', face.shape)   # (24, 21, 600)
```
